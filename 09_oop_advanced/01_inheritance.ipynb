## 1. Single Inheritance

````python
class College:
    def college_name(self):
        print("ABC College")
    def list_of_department(self):
        print(["CS","ECE", "MBA"])


class Student(College):
    pass

    # def student_name(self):
    #     print("Student Name: Arun")

s = Student()
s.college_name()
s.list_of_department()
````

## 2. Multiple Inheritance

````python
class Student:
    def student_details(self):
        print("Student Name: Arun")

class Management:
    def management_details(self):
        print("Management: MBA")

class CollegeStudent(Student, Management):
    def college_details(self):
        print("College: ABC College")



s = CollegeStudent()
s.student_details()
s.management_details()
s.college_details()
````

## 3. Multilevel Inheritance

````python
class College:
    def college_name(self):
        print("ABC College")

class Management(College):
    def management_name(self):
        print("Management Department")

class Student(Management):
    def student_name(self):
        print("Student Name: Arun")


class Visitor(Student):
    def visitor_name(self):
        print("visitor name")

s = Student()
s.college_name()
s.management_name()
s.student_name()
````

## 4. Hierarchical Inheritance

````python
class College:
    def college_name(self):
        print("ABC College")

class Student(College):
    def student_details(self):
        print("Student Name: Arun")

class Management(College):
    def management_details(self):
        print("Management Department")

s = Student()
s.college_name()
s.student_details()

m = Management()
m.college_name()
m.management_details()
````

## 5. Hybrid Inheritance

````python
class College:
    def college_name(self):
        print("ABC College")

class Student(College):
    def student_name(self):
        print("Student Name: Arun")

class Management(College):
    def management_name(self):
        print("Management Department")

class CollegeStudent(Student, Management):
    def course(self):
        print("Course: BCA")

s = CollegeStudent()
s.college_name()
s.student_name()
s.management_name()
s.course()
````

## Polymorphism

````python
class College:
    def college_location(self):
        print("Trichy")
    def college_name(self): # method overriding
        print("ABC College")


class Student(College):

    def student_details(self):
        print("Student Name: Arun")

    def college_name(self): # method overriding
        super().college_name()
        # print("xyz College")

s = Student()
s.college_name()
s.student_details()
````

## Summary

### Inheritance Syntax:

```python
class Parent:
    pass

class Child(Parent):  # Single inheritance
    pass

class Child(Parent1, Parent2):  # Multiple inheritance
    pass
```

### Key Concepts:

| Concept | Description |
|---------|-------------|
| Inheritance | Child gets all parent attributes/methods |
| Override | Child redefines parent method |
| `super()` | Call parent class methods |
| MRO | Method Resolution Order |

### Best Practices:
1. Use inheritance for "is-a" relationships
2. Always call `super().__init__()` in child
3. Prefer composition over deep inheritance
4. Use mixins for shared behavior
5. Keep inheritance hierarchies shallow
